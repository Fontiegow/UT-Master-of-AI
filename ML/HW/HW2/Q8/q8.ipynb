# Step 1: Loading and Exploring the MNIST Dataset

In this step, we load the MNIST dataset, which contains grayscale images of handwritten digits (0–9).
The goal is to inspect the structure of the data and understand its dimensions before any preprocessing
or modeling is performed.

MNIST consists of:
- 60,000 training images
- 10,000 test images
Each image is 28×28 pixels, represented as a grayscale matrix.


```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
```

```python
# Load MNIST dataset
(X_train_full, y_train_full), (X_test_full, y_test_full) = mnist.load_data()

print("Training set shape:", X_train_full.shape)
print("Training labels shape:", y_train_full.shape)
print("Test set shape:", X_test_full.shape)
print("Test labels shape:", y_test_full.shape)
```

```python
# Visualize some samples from the dataset
def plot_samples(X, y, num_samples=10):
    plt.figure(figsize=(10, 1))
    for i in range(num_samples):
        plt.subplot(1, num_samples, i + 1)
        plt.imshow(X[i], cmap='gray')
        plt.title(y[i])
        plt.axis('off')
    plt.show()
```

```python
# Display some random samples
plt.figure(figsize=(10, 4))
for i in range(10):
    idx = np.random.randint(0, len(X_train_full))
    plt.subplot(2, 5, i+1)
    plt.imshow(X_train_full[idx], cmap="gray")
    plt.title(f"Label: {y_train_full[idx]}")
    plt.axis("off")

plt.suptitle("Random Samples from MNIST Training Set")
plt.show()
```

From the visual inspection, we observe that:

- The input data are grayscale images.
- Digits are centered but not perfectly aligned.
- There is significant variation in handwriting styles.
- Pixel values range from 0 (black) to 255 (white).

This confirms that the problem is a non-trivial multi-class classification task,
suitable for testing neural networks.
